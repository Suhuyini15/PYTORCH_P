Q4 - RANDOM SEED

```python
# r_seed = 0
torch.manual_seed(0)
x = torch.rand(7,7)
y = torch.rand(1,7)
print(f'first tensor:\n{x}\n\n second tensor:\n{y}')
```

Q5 - RANDOM SEED (GPU)

```python
torch.manual_seed(1234)
x = torch.rand(7,7).cuda()
y = torch.rand(1,7).cuda()
print(f'first tensor:\n{x}\n\n second tensor:\n{y}')
#
```

Q6 - RANDOM TENSOR (GPU)

```python
torch.manual_seed(1234)
x = torch.rand(2,3).cuda()
y = torch.rand(2,3).cuda()
print(f'first tensor:\n{x}\n\n second tensor:\n{y}')
```

Q7 - MATMUL

```python
result = torch.matmul(x,y.T)
result
```

Q8 - MIN AND MAX VALUE

```python
result.max(), result.min()
```

Q9 - MIN AND MAX INDEX

```python
result.argmax(), result.argmin()
```
